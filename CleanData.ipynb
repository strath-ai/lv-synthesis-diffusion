```python
import numpy as np
train_samples = np.load("/home/abrash/LVGen/LVGen/nged_retriever/TrainingSamples.npy")
test_samples = np.load("/home/abrash/LVGen/LVGen/nged_retriever/testing_samples.npy")
```

```python
print(train_samples.shape, test_samples.shape)
print(max(train_samples[:, :,0].flatten()), max(test_samples[:, :,0].flatten()))
```

```python
def remove_extreme_samples(data, threshold=400):
    # Identify samples that contain extreme values
    mask = np.all(np.abs(data) < threshold, axis=(1, 2))  # shape: (num_samples,)
    return data[mask]

# Apply to both train and test samples
train_samples = train_samples[:, :, 0:2]
test_samples = test_samples[:, :, 0:2]

train_samples = remove_extreme_samples(train_samples)
test_samples = remove_extreme_samples(test_samples)
```

```python
print(train_samples.shape, test_samples.shape)
print(max(train_samples[:, :,0].flatten()), max(test_samples[:, :,0].flatten()))
```

```python
np.save("data/TrainingSamples_Cleaned.npy", train_samples)
np.save("data/testing_samples_Cleaned.npy", test_samples)
```
