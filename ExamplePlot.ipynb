```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

## Some exaples plots of data

NOTE: resutls are based on early checkpoints of models and do not represent models full performance capabilities

## Load data

```python
generated_data = np.load("Inference/results/LVGenWC/T200_beta00.0001_betaT0.02/imputation0.npy").transpose(0,2,1)
true_data = np.load("Inference/results/LVGenWC/T200_beta00.0001_betaT0.02/original0.npy").transpose(0,2,1)
```

## Plot generated vs real load profile

```python
#randomly select a substation to plot
substation_idx = np.random.randint(0, generated_data.shape[0])
plt.plot(generated_data[substation_idx,:,0], label="Generated Active Power")
plt.plot(true_data[substation_idx,:,0], label="True Active Power")
plt.legend()
plt.title("Generated vs True Active Power for Substation {}".format(substation_idx))
plt.xlabel("Time Steps")
plt.ylabel("Active Power")
plt.show()

#reactive power
plt.plot(generated_data[substation_idx,:,1], label="Generated Reactive Power")
plt.plot(true_data[substation_idx,:,1], label="True Reactive Power")
plt.legend()
plt.title("Generated vs True Reactive Power for Substation {}".format(substation_idx))
plt.xlabel("Time Steps")
plt.ylabel("Reactive Power")
plt.show()
```

# Compare Distribution

```python
#active power
plt.hist(generated_data[:,:,0].flatten(), bins=200, alpha=0.5, label="Generated Active Power")
plt.hist(true_data[:,:,0].flatten(), bins=200, alpha=0.5, label="True Active Power")
plt.legend()
plt.title("Distribution of Generated vs True Active Power for Substation {}".format(substation_idx))
plt.xlabel("Active Power")
plt.ylabel("Frequency")
plt.show()


#reactive power
plt.hist(generated_data[:,:,1].flatten(), bins=200, alpha=0.5, label="Generated Reactive Power")
plt.hist(true_data[:,:,1].flatten(), bins=200, alpha=0.5, label="True Reactive Power")
plt.legend()
plt.title("Distribution of Generated vs True Reactive Power for Substation {}".format(substation_idx))
plt.xlabel("Reactive Power")
plt.ylabel("Frequency")
plt.show()
```
